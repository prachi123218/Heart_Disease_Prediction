```python
from sklearn.preprocessing import LabelEncoder

le = LabelEncoder()

for col in ['gender', 'smoking', 'alcohol_consumption',
            'physical_activity', 'family_history', 'disease']:
    
    data[col] = le.fit_transform(data[col])
```
